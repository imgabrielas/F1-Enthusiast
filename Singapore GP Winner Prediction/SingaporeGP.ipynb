### How the weekend works
- 2026 is the first time Marina Bay hosts the Sprint format, and it is the last Sprint weekend of the season.
- There is only **one practice session**, so teams have just one hour of running before every session counts. Sprint Qualifying replaces FP2 and the Sprint replaces FP3.
- **Sprint Qualifying** (SQ1, SQ2, SQ3) sets the grid for the Sprint.
- The **Sprint** is a short race of about 100 km with no mandatory pit stop.
- **Qualifying** on Saturday night is separate from the Sprint and sets the grid for Sunday's Grand Prix.
- The **Race** is a night race over 62 laps of the 4.927 km Marina Bay Street Circuit.
